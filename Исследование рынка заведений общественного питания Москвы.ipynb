### Описание данных

Источники данных

Данные предоставлены в двух файлах и составлены на основе информации из сервисов Яндекс Карты и Яндекс Бизнес на лето 2022 года. Информация носит справочный характер и могла быть добавлена пользователями или найдена в общедоступных источниках.

Файл rest_info.csv — основная информация о заведениях

Этот файл содержит базовые характеристики заведений общественного питания Москвы и включает восемь столбцов.

Столбец name содержит название заведения. Столбец address содержит адрес заведения. Столбец district указывает административный район Москвы, например Центральный административный округ. Столбец category определяет категорию заведения, такую как кафе, ресторан, бар, пиццерия или кофейня. Столбец hours содержит информацию о днях и часах работы. Столбец rating хранит рейтинг заведения по оценкам пользователей в Яндекс Картах, где высшая оценка — 5.0. Столбец chain представляет собой числовой признак, выраженный 0 или 1, который показывает, является ли заведение сетевым. Значение 0 означает, что заведение не является сетевым, значение 1 — что является сетевым. При этом для маленьких сетей могут встречаться ошибки. Столбец seats содержит количество посадочных мест.

Файл rest_price.csv — ценовая информация

Этот файл содержит данные о ценах и средних чеках в заведениях и включает четыре столбца.

Столбец price содержит категорию цен в заведении, например средние, ниже среднего или выше среднего. Столбец avg_bill представляет собой строку, которая хранит среднюю стоимость заказа в виде диапазона. Примеры таких строк: «Средний счёт: 1000–1500 ₽», «Цена чашки капучино: 130–220 ₽» или «Цена бокала пива: 400–600 ₽». Столбец middle_avg_bill содержит число с оценкой среднего чека, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Средний счёт». Если в строке указан ценовой диапазон из двух значений, в этот столбец войдёт медиана этих двух значений. Если в строке указано одно число без диапазона, то в столбец войдёт это число. Если значения нет или оно не начинается с подстроки «Средний счёт», то в столбец ничего не войдёт, то есть будет пропуск. Столбец middle_coffee_cup содержит число с оценкой одной чашки капучино, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Цена одной чашки капучино». Принцип заполнения аналогичен: для диапазона берётся медиана, для одного числа — само число, в остальных случаях — пропуск.

Важные особенности данных, на которые стоит обратить внимание

Первая особенность связана с неполнотой ценовой информации. В столбце avg_bill могут быть указаны разные типы цен: средний чек, цена капучино, цена пива и другие. При этом поля middle_avg_bill и middle_coffee_cup заполнены выборочно и содержат много пропусков, так как заполняются только для соответствующих подстрок.

Вторая особенность — возможные ошибки в признаке chain. Для маленьких сетей могут встречаться неточности, поэтому к этому признаку следует относиться с осторожностью.

Третья особенность — наличие пропусков в рейтингах и количестве посадочных мест. Не все заведения имеют оценку пользователей, и информация о посадочных местах также может отсутствовать.

Четвёртая особенность — текстовые поля требуют нормализации. Названия районов и категорий могут быть записаны в разных форматах, а часы работы требуют парсинга для выделения круглосуточных заведений.

Пятая особенность — объединение датасетов. Датасеты не имеют явного общего ключа, поэтому объединение будет выполняться по индексам, то есть по порядку строк.

```python
print("=" * 60)
print("ПРОВЕРКА ДУБЛИКАТОВ")
print("=" * 60)

# Функция нормализации
def normalize_text(text):
    if pd.isna(text):
        return text
    text = str(text).strip().lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[.,;:!?()"]', '', text)
    return text

# Создаём нормализованные столбцы
df['name_norm'] = df['name'].apply(normalize_text)
df['address_norm'] = df['address'].apply(normalize_text)

# Полные дубликаты
full_dups = df.duplicated().sum()
print(f"Полных дубликатов строк: {full_dups}")

# Дубликаты по названию и адресу
name_addr_dups = df.duplicated(subset=['name_norm', 'address_norm']).sum()
print(f"Дубликатов по названию и адресу: {name_addr_dups}")

# Удаляем только полные дубликаты (НЕ удаляем дубликаты по названию и адресу,
# так как это могут быть разные заведения с одинаковым названием или адресом)
df_len_before = len(df)
df = df.drop_duplicates()
df_len_after = len(df)
print(f"\nУдалено полных дубликатов: {df_len_before - df_len_after}")

# Удаляем временные столбцы
df = df.drop(['name_norm', 'address_norm'], axis=1)

print(f"\nРазмер данных после удаления полных дубликатов: {df.shape[0]} строк")
```

```python
print("=" * 60)
print("СОЗДАНИЕ СТОЛБЦА is_24_7")
print("=" * 60)

def is_24_7(hours_str):
    if pd.isna(hours_str):
        return False
    hours_str = str(hours_str).lower()
    if '24' in hours_str and ('ч' in hours_str or 'час' in hours_str):
        return True
    if 'круглосуточно' in hours_str:
        return True
    if re.search(r'00:00\s*[-–]\s*24:00', hours_str):
        return True
    if re.search(r'00:00\s*[-–]\s*23:59', hours_str):
        return True
    return False

df['is_24_7'] = df['hours'].apply(is_24_7)

print(f"Круглосуточных заведений: {df['is_24_7'].sum()} ({df['is_24_7'].sum()/len(df)*100:.1f}%)")
```

```python
print("=" * 70)
print("ПРОВЕРКА ДУБЛИКАТОВ С НОРМАЛИЗАЦИЕЙ ТЕКСТОВЫХ ПОЛЕЙ")
print("=" * 70)

# 1. Функция нормализации текста
def normalize_text(text):
    """
    Нормализует текст для поиска дубликатов:
    - приводит к нижнему регистру
    - удаляет пробелы в начале и конце
    - заменяет множественные пробелы на один
    - удаляет пунктуацию
    """
    if pd.isna(text):
        return text
    text = str(text).strip()
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[.,;:!?()"\'«»]', '', text)
    text = re.sub(r'[–—]', '-', text)
    return text

# 2. Создаем нормализованные столбцы
df['name_norm'] = df['name'].apply(normalize_text)
df['address_norm'] = df['address'].apply(normalize_text)
df['category_norm'] = df['category'].apply(normalize_text)
df['district_norm'] = df['district'].apply(normalize_text)

print("✓ Созданы нормализованные столбцы")
print("\nПример нормализации:")
print(f"Оригинальное название: {df['name'].iloc[0]}")
print(f"Нормализованное название: {df['name_norm'].iloc[0]}")
print(f"Оригинальный адрес: {df['address'].iloc[0]}")
print(f"Нормализованный адрес: {df['address_norm'].iloc[0]}")
```

```python
# Создаём функцию для определения круглосуточных заведений
def is_24_7(hours_str):
    """
    Определяет, работает ли заведение круглосуточно (24/7).
    
    Параметры:
    hours_str (str): строка с информацией о часах работы
    
    Возвращает:
    bool: True если заведение работает круглосуточно, иначе False
    """
    # Если данные отсутствуют, возвращаем False
    if pd.isna(hours_str):
        return False
    
    # Приводим к строке и нижнему регистру для удобства поиска
    hours_str = str(hours_str).lower()
    
    # Проверяем наличие ключевых слов и фраз
    # 1. Прямое указание "круглосуточно"
    if 'круглосуточно' in hours_str:
        return True
    
    # 2. Указание "24 часа" или "24 ч"
    if '24' in hours_str and ('ч' in hours_str or 'час' in hours_str):
        return True
    
    # 3. Формат "00:00-24:00"
    if re.search(r'00:00\s*[-–]\s*24:00', hours_str):
        return True
    
    # 4. Формат "00:00-23:59"
    if re.search(r'00:00\s*[-–]\s*23:59', hours_str):
        return True
    
    # 5. Если не подходит ни под одно условие, возвращаем False
    return False

# Применяем функцию ко всем записям
df['is_24_7'] = df['hours'].apply(is_24_7)

print("✓ Столбец is_24_7 создан")
print(f"Количество круглосуточных заведений: {df['is_24_7'].sum()}")
print(f"Доля круглосуточных заведений: {df['is_24_7'].sum() / len(df) * 100:.1f}%")
```

```python
print("=" * 70)
print("АНАЛИЗ РАЗЛИЧИЙ РЕЙТИНГОВ МЕЖДУ КАТЕГОРИЯМИ")
print("=" * 70)

# Расчет относительных различий
overall_mean = rating_df['rating'].mean()
rating_by_cat['diff_from_mean'] = rating_by_cat['mean'] - overall_mean
rating_by_cat['diff_pct'] = (rating_by_cat['diff_from_mean'] / overall_mean) * 100

print("Отклонение среднего рейтинга каждой категории от общего среднего:")
display(rating_by_cat[['mean', 'diff_from_mean', 'diff_pct']].sort_values('mean', ascending=False))

# Проверка статистической значимости различий (ANOVA)
from scipy import stats

# Группируем рейтинги по категориям
groups = [rating_df[rating_df['category'] == cat]['rating'].values for cat in rating_by_cat.index]

# Проводим ANOVA-тест
f_stat, p_value = stats.f_oneway(*groups)

print(f"\nРЕЗУЛЬТАТЫ СТАТИСТИЧЕСКОГО ТЕСТА (ANOVA):")
print(f"  F-статистика: {f_stat:.4f}")
print(f"  p-value: {p_value:.6f}")

if p_value < 0.05:
    print("  ✓ Различия между средними рейтингами категорий СТАТИСТИЧЕСКИ ЗНАЧИМЫ (p < 0.05)")
    print("    → Но практическая значимость может быть невелика")
else:
    print("  ✗ Различия между средними рейтингами категорий НЕ ЗНАЧИМЫ (p >= 0.05)")
    print("    → Средние рейтинги категорий статистически не различаются")
```
